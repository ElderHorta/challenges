```python
import pandas as pd
from pandasql import sqldf
```

```python
pysqldf = lambda q: sqldf(q, globals())
```

```python
buyers = pd.read_csv('./content/buyers.csv', sep=',')
order_items = pd.read_csv('./content/order_items.csv', sep=',')
orders = pd.read_csv('./content/orders.csv', sep=',')
payments = pd.read_csv('./content/payments.csv', sep=',')
products = pd.read_csv('./content/products.csv', sep=',')
sellers = pd.read_csv('./content/sellers.csv', sep=',')
```

```python

def quality_check(df: pd.DataFrame, df_name: str = "DataFrame") -> pd.DataFrame:
    """
    Realiza verificações de qualidade genéricas em um DataFrame.

    Checks realizados por coluna:
      - Qtd. de nulos e percentual
      - Tipo de dado
      - Para colunas numéricas: valores negativos (qtd. e %)
      - Qtd. de valores únicos e percentual de unicidade
      - Aviso se coluna parece ser um identificador único
      - Aviso se coluna tem todos os valores duplicados (cardinalidade 1)
    """
    rows = []
    n = len(df)

    for col in df.columns:
        serie = df[col]

        nulls      = int(serie.isna().sum())
        null_pct   = round(nulls / n * 100, 2) if n > 0 else 0.0

        dtype = str(serie.dtype)

        if pd.api.types.is_numeric_dtype(serie):
            negatives     = int((serie < 0).sum())
            negative_pct  = round(negatives / n * 100, 2) if n > 0 else 0.0
        else:
            negatives    = None
            negative_pct = None

        n_unique      = int(serie.nunique(dropna=False))
        unique_pct    = round(n_unique / n * 100, 2) if n > 0 else 0.0
        is_unique_id  = (n_unique == n)
        all_duplicate = (n_unique == 1) and (n > 1)

        issues = []
        if nulls > 0:
            issues.append(f"⚠️ {null_pct}% nulos")
        if negatives:
            issues.append(f"⚠️ {negative_pct}% negativos")
        if all_duplicate:
            issues.append("⚠️ Coluna constante (1 valor único)")
        if is_unique_id:
            issues.append("✅ Identificador único (100% distinct)")

        rows.append({
            "coluna"           : col,
            "tipo"             : dtype,
            "total_linhas"     : n,
            "nulos"            : nulls,
            "nulos_%"          : null_pct,
            "negativos"        : negatives,
            "negativos_%"      : negative_pct,
            "valores_unicos"   : n_unique,
            "unicidade_%"      : unique_pct,
            "eh_id_unico"      : is_unique_id,
            "coluna_constante" : all_duplicate,
            "observacoes"      : " | ".join(issues) if issues else "✅ OK",
        })

    result = pd.DataFrame(rows)
    print(f"\n{'='*60}")
    print(f"  QUALITY CHECK → {df_name}  ({n} linhas × {len(df.columns)} colunas)")
    print(f"{'='*60}\n")
    display(result)
    return result


quality_check(buyers, df_name="buyers")
quality_check(order_items, df_name="order_items")
quality_check(orders, df_name="orders")
quality_check(payments, df_name="payments")
quality_check(products, df_name="products")
quality_check(sellers, df_name="sellers")
```
