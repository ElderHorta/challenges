```python
#O time financeiro reclama que o total de faturamento mensal varia dependendo de quem puxa o
#relatório. Você descobre que pedidos com status 'cancelled' e 'refunded' estão sendo incluídos em
#alguns relatórios.
#
#Escreva uma query que retorne o faturamento bruto mensal dos últimos 12 meses,
#considerando apenas pedidos com status 'completed' ou 'delivered'. Inclua também a
#quantidade de pedidos e o ticket médio de cada mês. Ordene do mês mais recente ao mais
#antigo.

query_desafio1 = """
WITH pedidos_validos AS (
    SELECT
        id,
        status,
        created_at,
        total_value
    FROM orders
    WHERE status IN ('completed', 'delivered')
    GROUP BY id
    HAVING created_at = MAX(created_at)
)
SELECT
    STRFTIME('%Y-%m', ped.created_at) AS ano_mes,
    PRINTF('%,.2f', SUM(ped.total_value)) AS faturamento_bruto,
    PRINTF('%,.2f', COUNT(*)) AS qtd_pedidos,
    PRINTF('%,.2f', ROUND(SUM(ped.total_value) / COUNT(*), 2)) AS ticket_medio
FROM pedidos_validos ped
WHERE ped.created_at >= (
    SELECT DATE(MAX(created_at), 'start of month', '-11 months')
    FROM orders
)
GROUP BY ano_mes
ORDER BY ano_mes DESC
"""

resultado_desafio1 = pysqldf(query_desafio1)
display(resultado_desafio1)
```

```python
#A diretora comercial quer um ranking dos 10 sellers com maior crescimento de GMV entre o
#trimestre atual e o anterior, mas só quer ver sellers que tiveram pelo menos 50 pedidos em ambos
#os trimestres (para evitar distorção de sellers novos ou inativos).

#Escreva a query que resolve esse problema, exibindo: nome do seller, estado, GMV do
#trimestre anterior, GMV do trimestre atual e o percentual de crescimento. Ordene pelo maior
#crescimento.


query_desafio2 = """
WITH pedidos_validos AS (
    SELECT
        id,
        seller_id,
        created_at,
        total_value,
        -- Identifica o trimestre: atual = últimos 3 meses, anterior = 3 meses antes
        CASE
            WHEN created_at >= DATE(max_date, 'start of month', '-2 months', 'start of month')
                THEN 'atual'
            WHEN created_at >= DATE(max_date, 'start of month', '-5 months', 'start of month')
             AND created_at < DATE(max_date, 'start of month', '-2 months', 'start of month')
                THEN 'anterior'
        END AS trimestre
    FROM orders,
         (SELECT MAX(created_at) AS max_date FROM orders) ref
    WHERE status IN ('completed', 'delivered')
    GROUP BY id
    HAVING created_at = MAX(created_at)
),
gmv_por_seller AS (
    SELECT
        seller_id,
        SUM(CASE WHEN trimestre = 'anterior' THEN total_value END) AS gmv_anterior,
        SUM(CASE WHEN trimestre = 'atual' THEN total_value END) AS gmv_atual,
        SUM(CASE WHEN trimestre = 'anterior' THEN 1 ELSE 0 END) AS pedidos_anterior,
        SUM(CASE WHEN trimestre = 'atual' THEN 1 ELSE 0 END) AS pedidos_atual
    FROM pedidos_validos
    WHERE trimestre IS NOT NULL
    GROUP BY seller_id
    HAVING pedidos_anterior >= 50 AND pedidos_atual >= 50
)
SELECT
    sel.name AS seller,
    sel.state AS estado,
    PRINTF('%,.2f', ROUND(gmv.gmv_anterior, 2)) AS gmv_trimestre_anterior,
    PRINTF('%,.2f', ROUND(gmv.gmv_atual, 2)) AS gmv_trimestre_atual,
    ROUND((gmv.gmv_atual - gmv.gmv_anterior) / gmv.gmv_anterior * 100, 2) AS crescimento_pct
FROM gmv_por_seller gmv
JOIN sellers sel ON sel.id = gmv.seller_id
ORDER BY (gmv.gmv_atual - gmv.gmv_anterior) / gmv.gmv_anterior * 100 DESC
LIMIT 10
"""

resultado_desafio2 = pysqldf(query_desafio2)
display(resultado_desafio2)
```

```python
#O time de fraudes suspeita que alguns sellers estão aplicando descontos abusivos para inflar
#volume artificialmente. Você precisa encontrar todos os pedidos onde o desconto total (soma dos
#descontos dos itens) representa mais de 40% do valor bruto do pedido, listando também o seller
#responsável e a data do pedido. Exclua pedidos cancelados.

query_desafio3 = """
WITH itens_por_pedido AS (
    SELECT
        order_id,
        SUM(qty * unit_price) AS valor_bruto,
        SUM(discount) AS desconto_total
    FROM order_items
    GROUP BY order_id
    HAVING SUM(qty * unit_price) > 0
       AND SUM(discount) / SUM(qty * unit_price) > 0.40
),
pedidos_dedup AS (
    SELECT id, seller_id, status, created_at
    FROM orders
    WHERE status != 'cancelled'
    GROUP BY id
    HAVING created_at = MAX(created_at)
)
SELECT
    ord.id AS order_id,
    sel.name AS seller,
    DATE(ord.created_at) AS data_pedido,
    PRINTF('%,.2f', ROUND(itp.valor_bruto, 2)) AS valor_bruto,
    PRINTF('%,.2f', ROUND(itp.desconto_total, 2)) AS desconto_total,
    ROUND(itp.desconto_total / itp.valor_bruto * 100, 2) AS desconto_pct,
    ord.status
FROM pedidos_dedup ord
JOIN itens_por_pedido itp ON itp.order_id = ord.id
JOIN sellers sel ON sel.id = ord.seller_id
ORDER BY desconto_pct DESC
"""

resultado_desafio3 = pysqldf(query_desafio3)
print(f"Total de pedidos com desconto > 40%: {len(resultado_desafio3)}")
display(resultado_desafio3)
```

```python
#Existe um produto com comportamento estranho: ele tem um volume de vendas alto, mas nunca
#aparece como item mais vendido dentro de nenhum pedido (nunca é o item de maior valor num
#pedido). Encontre todos os produtos que se encaixam nessa descrição: total de unidades vendidas
#> 1.000, mas que em nenhum pedido foram o item de maior valor unitário.

#Dica: window functions podem ser suas aliadas aqui e avalie os resultados e viabilidade da
#análise e faça os questionamentos necessários.
```

```python
# Interpretação: "item de maior valor num pedido" = item que mais contribuiu para o
# total do pedido, considerando quantidade, preço negociado e desconto aplicado.
# Fórmula: qty * unit_price (valor líquido do item naquele pedido supondo que o desconto já foi aplicado no unit_price)

query_desafio4_valor_item = """
WITH volume_total AS (
    SELECT product_id, SUM(qty) AS total_unidades
    FROM order_items
    GROUP BY product_id
    HAVING SUM(qty) > 1000
),
pedidos_multi_item AS (
    SELECT order_id
    FROM order_items
    GROUP BY order_id
    HAVING COUNT(*) > 1
),
ranked AS (
    SELECT
        oit.product_id,
        RANK() OVER (
            PARTITION BY oit.order_id
            ORDER BY (oit.qty * oit.unit_price) DESC
        ) AS rank_valor
    FROM order_items oit
    JOIN pedidos_multi_item mul ON mul.order_id = oit.order_id
),
stats_por_produto AS (
    SELECT
        product_id,
        COUNT(*) AS aparicoes_multi_item,
        SUM(CASE WHEN rank_valor = 1 THEN 1 ELSE 0 END) AS vezes_top1
    FROM ranked
    GROUP BY product_id
)
SELECT
    pro.id AS product_id,
    pro.name AS produto,
    pro.category AS categoria,
    vol.total_unidades,
    sta.aparicoes_multi_item,
    sta.vezes_top1
FROM stats_por_produto sta
JOIN products pro ON pro.id = sta.product_id
INNER JOIN volume_total vol ON vol.product_id = sta.product_id
WHERE sta.vezes_top1 = 0
ORDER BY vol.total_unidades DESC
"""

resultado_desafio4_valor_item = pysqldf(query_desafio4_valor_item)
print(f"Produtos com > 1.000 un. vendidas e nunca top 1 por valor total do item: {len(resultado_desafio4_valor_item)}")
display(resultado_desafio4_valor_item)
```

```python
# Versão alternativa: usando unit_cost * qty - discount como "maior valor no pedido"
#
# unit_cost (products) = custo fixo de catálogo do produto
# Fórmula: qty * unit_cost - discount (valor de custo líquido do item naquele pedido)

query_desafio4_custo_item = """
WITH volume_total AS (
    SELECT product_id, SUM(qty) AS total_unidades
    FROM order_items
    GROUP BY product_id
    HAVING SUM(qty) > 1000
),
pedidos_multi_item AS (
    SELECT order_id
    FROM order_items
    GROUP BY order_id
    HAVING COUNT(*) > 1
),
ranked AS (
    SELECT
        oit.product_id,
        RANK() OVER (
            PARTITION BY oit.order_id
            ORDER BY (oit.qty * pro.unit_cost - oit.discount) DESC
        ) AS rank_valor
    FROM order_items oit
    JOIN products pro ON pro.id = oit.product_id
    JOIN pedidos_multi_item mul ON mul.order_id = oit.order_id
),
stats_por_produto AS (
    SELECT
        product_id,
        COUNT(*) AS aparicoes_multi_item,
        SUM(CASE WHEN rank_valor = 1 THEN 1 ELSE 0 END) AS vezes_top1
    FROM ranked
    GROUP BY product_id
)
SELECT
    pro.id AS product_id,
    pro.name AS produto,
    pro.category AS categoria,
    ROUND(pro.unit_cost, 2) AS unit_cost,
    vol.total_unidades,
    sta.aparicoes_multi_item,
    sta.vezes_top1
FROM stats_por_produto sta
JOIN products pro ON pro.id = sta.product_id
INNER JOIN volume_total vol ON vol.product_id = sta.product_id
WHERE sta.vezes_top1 = 0
ORDER BY vol.total_unidades DESC
"""

resultado_desafio4_custo_item = pysqldf(query_desafio4_custo_item)
print(f"Produtos com > 1.000 un. vendidas e nunca top 1 por (qty * unit_cost - discount): {len(resultado_desafio4_custo_item)}")
display(resultado_desafio4_custo_item)
```
